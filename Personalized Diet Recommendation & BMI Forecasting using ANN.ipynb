```python
import pandas as pd
import numpy as np


import matplotlib.pyplot as plt


from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten
import warnings
warnings.filterwarnings('ignore')
import joblib

Personalized_Diet_Recommendations=pd.read_csv(r"C:\Users\Afaq_Lite\Downloads\Personalized_Diet_Recommendations.csv")

Personalized_Diet_Recommendations

Personalized_Diet_Recommendations.info()

forecasting_table=pd.read_csv(r"C:\Users\Afaq_Lite\Downloads\forecasting.csv")
forecasting_table

forecasting_table.info()

fill_none_cols = ['Chronic_Disease', 'Allergies', 'Food_Aversions']
for col in fill_none_cols:
    if col in Personalized_Diet_Recommendations.columns:
        Personalized_Diet_Recommendations[col] = Personalized_Diet_Recommendations[col].fillna('None')


Personalized_Diet_Recommendations.duplicated().sum()

forecasting_table.duplicated().sum()

unique_Personalized_Diet_Recommendations= [
    'Patient_ID', 'Blood_Sugar_Level', 'Genetic_Risk_Factor', 'Allergies', 
    'Daily_Steps', 'Exercise_Frequency', 'Sleep_Hours', 'Alcohol_Consumption', 
    'Smoking_Habit', 'Dietary_Habits', 'Caloric_Intake', 'Protein_Intake', 
    'Carbohydrate_Intake', 'Fat_Intake', 'Preferred_Cuisine', 'Food_Aversions'
]
df_Personalized_Diet_Recommendations = Personalized_Diet_Recommendations[unique_Personalized_Diet_Recommendations]


data = pd.merge(forecasting_table, df_Personalized_Diet_Recommendations , on="Patient_ID", how="left")


data.head()

data.info()

data.duplicated().sum()

object_col=["Patient_ID","Gender","Chronic_Disease","Recommended_Meal_Plan","Genetic_Risk_Factor","Allergies","Alcohol_Consumption",
            "Smoking_Habit","Dietary_Habits","Preferred_Cuisine","Food_Aversions"]
for i in object_col:
    Dict={k:j for  k,j in zip (data[i].unique(),range(data[i].nunique()))}
    data[i]=data[i].map(Dict)


data.info()

data = data.sort_values(by=['Patient_ID', 'Time_Step'])


exclude_cols = ['Patient_ID', 'Time_Step']
features_list = [col for col in data.columns if col not in exclude_cols]


scaler = StandardScaler()
data_scaled = data.copy()
data_scaled[features_list] = scaler.fit_transform(data[features_list])

num_patients = data['Patient_ID'].nunique()   
num_timesteps = 12                           
num_features = len(features_list)              
data_3d = data_scaled[features_list].values.reshape(num_patients, num_timesteps, num_features)

bmi_index = features_list.index('BMI')
X = data_3d[:, :-1, :]   
y = data_3d[:, -1, bmi_index]  

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=21)

model = Sequential([
    Flatten(input_shape=(X_train.shape[1], X_train.shape[2])),
    Dense(8, activation='relu'),
    Dense(128, activation='relu'),
    #Dropout(0.2),
    Dense(64, activation='relu'),
    #Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(1)])
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()

history = model.fit(
    X_train, y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1)
test_loss, test_mae = model.evaluate(X_test, y_test, verbose=0)
print(f"BMI with MAE: {test_mae:.4f}")

plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Validation loss Curve')
plt.xlabel('Epochs')
plt.ylabel('MSE')
plt.legend()
plt.grid(True)
plt.show()

y_pred_scaled = model.predict(X_test)

dummy_pred = np.zeros((len(y_test), num_features))
dummy_real = np.zeros((len(y_test), num_features))

dummy_pred[:, bmi_index] = y_pred_scaled.flatten()
dummy_real[:, bmi_index] = y_test
y_pred_original = scaler.inverse_transform(dummy_pred)[:, bmi_index]
y_real_original = scaler.inverse_transform(dummy_real)[:, bmi_index]

comparison_df = pd.DataFrame({
    "BMI_Actual": y_real_original[:10],
     "BMI_Predicted": y_pred_original[:10],
    'Absolute Error': np.abs(y_real_original[:10] - y_pred_original[:10])})

print("Actual and Predicted table")
display(comparison_df)

model.save('bmi_ann_model.keras')

joblib.dump(scaler, 'data_scaler.pkl')



def generate_diet_recommendation(row):
    bmi = row['BMI']
    disease = row['Chronic_Disease']
    if bmi < 18.5:
        weight_status = "Underweight"
        caloric_action = "Increase caloric intake to build muscle mass."
    elif 18.5 <= bmi < 25:
        weight_status = "Normal Weight"
        caloric_action = "Maintain current caloric intake to stabilize weight."
    else:
        weight_status = "Overweight/Obesity"
        caloric_action = "Reduce caloric intake to assist in gradual weight loss."
    recommendation = f"Health Status: {weight_status}. Action Plan: {caloric_action} "
    
    if 'diabet' in str(disease).lower() or disease == 1:
        recommendation += "Dietary Advice: Follow a Low-Carb diet and reduce sugar intake to help regulate blood glucose levels."
    elif 'hyper' in str(disease).lower() or disease == 2:
        recommendation += "Dietary Advice: Follow a Low-Fat diet and limit sodium consumption to help stabilize blood pressure."
    elif 'obese' in str(disease).lower() or disease == 3:
        recommendation += "Dietary Advice: Prioritize high-fiber and high-protein foods to enhance satiety and promote fat burning."
    else:
        recommendation += "Dietary Advice: Follow a well-balanced diet containing all essential macronutrients."
        
    return recommendation
recommendations_sample = data[data['Time_Step'] == 0].head(5).copy()
recommendations_sample['Diet_Recommendation'] = recommendations_sample.apply(generate_diet_recommendation, axis=1)


print("\n*** Sample Diet Recommendations for Distinct Patients ***")
for idx, row in recommendations_sample.iterrows():
    print(f"\n Patient ID {int(row['Patient_ID'])}:")
    print(f" Predicted BMI: {row['BMI']:.2f}")
    print(f"Recommendation: {row['Diet_Recommendation']}")

```
